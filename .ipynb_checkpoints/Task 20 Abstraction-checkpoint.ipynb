```python
# Part A – Abstraction Without super() (1–15)
# 1. Class Name: Shape
# Problem
# Create an abstract class named Shape.
# Explanation
# A shape can have different formulas for area and perimeter. Therefore, create two abstract
# methods:
# • area()
# • perimeter()
# Create subclasses like Circle and Rectangle and implement both methods.

from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

# Child Class 1
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14 * self.radius

# Child Class 2
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
        
    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)
        
# Objects
c = Circle(9)
print("Circle")
print("Area:", c.area())
print("Perimeter:", c.perimeter())
print()

r = Rectangle(10, 5)
print("Rectangle")
print("Area:", r.area())
print("Perimeter:", r.perimeter())
```

```python
# 2. Class Name: Vehicle
# Problem
# Create an abstract class named Vehicle.
# Explanation
# Every vehicle can start and stop, but each vehicle does it differently.
# Create two abstract methods:
# • start()
# • stop()
# Create a Car class that implements both methods.


from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

# Child Class 1
class Car (Vehicle):
    def start(self):
        print("Car start")
        
    def stop(self):
        print("Stop car")

c = Car()
c.start()
c.stop()
```

```python
# 3. Class Name: Animal
# Problem
# Create an abstract class named Animal.
# Explanation
# Every animal makes a sound, but the sound depends on the animal.
# Create an abstract method:
# • make_sound()
# Create subclasses:
# • Dog
# • Cat
# Each should implement its own sound.

from abc import ABC, abstractmethod

# Abstract Class
class Animal (ABC):

    @abstractmethod
    def make_sound(self):
        pass

# Child Class 1
class Dog (Animal):
    def make_sound(self):
        print("bhowwwww")
# Child Class 1
class Cat (Animal):
    def make_sound(self):
        print("Meaoooooo")

d = Dog()
d.make_sound()
c = Cat()
c.make_sound()
```

```python
# 4. Class Name: Payment
# Problem
# Create an abstract class named Payment.
# Explanation
# Every payment processes money differently.
# Create an abstract method:
# • process_payment(amount)
# Create subclasses:
# • CreditCard
# • Cash
# Each should process the payment in its own way.

from abc import ABC, abstractmethod

# Abstract Class
class Payment (ABC):

    @abstractmethod
    def process_payment(self,amount):
        pass

# Child Class 1
class Credit_card (Payment):
    def process_payment(self,amount):
        print(f"Credit card Payment : {amount}")
# Child Class 2
class Cash (Payment):
    def process_payment(self,amount):
        print(f"Payment Cash : {amount}")

credit = Credit_card()
cash = Cash()
credit.process_payment(5000)
cash.process_payment(10000)
```

```python
# 9. Class Name: Transport
# Problem
# Create an abstract class named Transport.
# Explanation
# Every transport moves differently and uses different fuel.
# Create abstract methods:
# • move()
# • fuel_type()
# Create subclasses like:
# • Car
# • Train


from abc import ABC, abstractmethod

# Abstract Class
class Transport(ABC):

    @abstractmethod
    def move(self):
        pass

    @abstractmethod
    def fuel_type(self):
        pass

# Child Class 1
class Car(Transport):
    def move(self):
        print("Car move on road")
        
    def fuel_type(self):
        print("Petrol/Diesel/Electric")

# Child Class 2
class Train(Transport):
    def move(self):
        print("Train moves on railway tracks")
        
    def fuel_type(self):
        print("Diesel/Electric")

car = Car()
car.move()
car.fuel_type()

print()
train = Train()
train.move()
train.fuel_type()
```

```python
# 10. Class Name: Notification
# Problem
# Create an abstract class named Notification.
# Explanation
# Notifications can be sent in different ways.
# Create abstract method:
# • send(message)
# Create subclasses:
# • EmailNotification
# • SMSNotification

from abc import ABC, abstractmethod

# Abstract Class
class Notification (ABC):

    @abstractmethod
    def send(self,message):
        pass


# Child Class 1
class Email(Notification):
    def send(self, message):
        print("Email Notification")
        print("Message:", message)

# Child Class 2
class SMS(Notification):
    def send(self, message):
        print("SMS Notification")
        print("Message:", message)

email = Email()
sms = SMS()
email.send("Your order has been shipped")
print()

sms.send("Your Msg")
```

```python
# 11. Class Name: FileHandler
# Problem
# Create an abstract class named FileHandler.
# Explanation
# Files can be read and written differently.
# Create abstract methods:
# • read()
# • write(data)
# Create subclasses like:
# • TextFileHandler
# • CSVFileHandler

from abc import ABC, abstractmethod

# Abstract Class
class File_Handler(ABC):

    @abstractmethod
    def read(self):
        pass

    @abstractmethod
    def write(self, data):
        pass

# Child Class 1
class TextFileHandler(File_Handler):
    def read(self):
        print("Text file handler can be raed")

        
    def write(self, data):
        self.data = data
        print(f"write the data in Text file : {self.data}")

# Child Class 2
class CSVFileHandler(File_Handler):
    def read(self):
        print("CSV file can be read")
        
    def write(self, data):
        self.data = data
        print(f"write the data in csv file : {self.data}")

text = TextFileHandler()
text.read()
text.write("Hello, World!")

print()
csv = CSVFileHandler()
csv.read()
csv.write("sahil vaidya age:23")
```

```python
# 13. Class Name: Calculator
# Problem
# Create an abstract class named Calculator.
# Explanation
# Different calculators perform different operations.
# Create abstract method:
# • compute(a, b)
# Create subclasses:
# • AdditionCalculator
# • MultiplicationCalculator

from abc import ABC, abstractmethod

# Abstract Class
class Calculator(ABC):

    @abstractmethod
    def compute(self,a,b):
        pass

# Child Class 1
class AdditionCalculator(Calculator):
    def compute(self,a,b):
        total = a + b
        print(f"Addition calculator       : {total}")   

# Child Class 2
class MultiplicationCalculator(Calculator):
    def compute(self, a, b):
        total = a * b
        print(f"Multiplication Calculator : {total}")

add = AdditionCalculator()
mult = MultiplicationCalculator()

add.compute(20,30)
mult.compute(5,5)
```

```python
# 14. Class Name: Database
# Problem
# Create an abstract class named Database.
# Explanation
# Different databases connect and execute queries differently.
# Create abstract methods:
# • connect()
# • query(sql)
# Create subclasses:
# • MySQLDatabase
# • MongoDatabase


from abc import ABC, abstractmethod

# Abstract Class
class Database(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def query(self, sql):
        pass

# Child Class 1
class MySQLDatabase(Database):
    def connect(self):
        print("MY SQL Database is connect")

    def query(self, sql):
        self.sql = sql
        print(f"MY SQL Database Query : {self.sql}")

# Child Class 2
class MongoDatabase(Database):
    def connect(self):
        print("Mongo Data Base is connected ")
        
    def query(self, sql):
        self.sql = sql
        print(f"mongo Data Base Query : {self.sql}")

mysql = MySQLDatabase()
mongo = MongoDatabase()

print("MySQL Database:")
mysql.connect()
mysql.query("SELECT * FROM employees")

print()

print("MongoDB Database:")
mongo.connect()
mongo.query("db.students.find()")
```

```python
# 15. Class Name: Shape3D
# Problem
# Create an abstract class named Shape3D.
# Explanation
# Three-dimensional shapes have different formulas.
# Create abstract methods:
# • volume()
# • surface_area()
# Create subclasses:
# • Cube
# • Sphere

from abc import ABC, abstractmethod

# Abstract Class
class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass

# Child Class 1
class Cube(Shape3D):
    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * (self.side ** 2)

    def display(self):
        print("Cube")
        print("Side         :", self.side)
        print("Volume       :", self.volume())
        print("Surface Area :", self.surface_area())
        
# Child Class 2
class Sphere(Shape3D):
    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4/3) * 3.14 * (self.radius ** 3)

    def surface_area(self):
        return 4 * 3.14 * (self.radius ** 2)

    def display(self):
        print("Sphere")
        print("Radius       :", self.radius)
        print("Volume       :", self.volume())
        print("Surface Area :", self.surface_area())
        
# Objects
cube = Cube(4)
sphere = Sphere(3)

cube.display()
print()

sphere.display()
```

```python
# Part B – Abstraction Using super() (16–30)
# 16. Class Name: Shape
# Problem
# Create an abstract class named Shape.
# Explanation
# The constructor should accept a shape's color.
# __init__(self, color)
# Create an abstract method:
# • area()
# Create subclass:
# • Circle
# Use super() to initialize the color.

from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):

    def __init__(self, color):
        self.color = color

    @abstractmethod
    def area(self):
        pass

# Child Class
class Circle(Shape):

    def __init__(self, color, radius):
        super().__init__(color)      # Call parent constructor
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

    def display(self):
        print("Shape  :", "Circle")
        print("Color  :", self.color)
        print("Radius :", self.radius)
        print("Area   :", self.area())

# Object
c1 = Circle("Red", 5)
c1.display()
```

```python
# 17. Class Name: Vehicle
# Problem
# Create an abstract class named Vehicle.
# Explanation
# Store the vehicle brand in the constructor.
# __init__(self, brand)
# Create abstract methods:
# • start()
# • stop()
# Create subclass:
# • Car
# Use super() to initialize the brand.


from abc import ABC, abstractmethod

# Abstract Class
class Vehicle(ABC):

    def __init__(self, brand):
        self.brand = brand

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

# Child Class
class Car(Vehicle):

    def __init__(self, brand, model):
        super().__init__(brand)      # Call parent constructor
        self.model = model

    def start(self):
        print("Car has started.")

    def stop(self):
        print("Car has stopped.")

    def display(self):
        print("Car:")
        print("Brand :",self.brand)
        print("Model :",self.model)
        self.start
        self.stop

# Object
car1 = Car("Toyota", "Fortuner")
car1.display()
```

```python
# 19. Class Name: Payment
# Problem
# Create an abstract class named Payment.
# Explanation
# Store the payment amount in the constructor.
# __init__(self, amount)
# Create abstract method:
# • process_payment()
# Create subclasses:
# • CreditCard
# • Cash
# Use super() to initialize the amount.

from abc import ABC, abstractmethod

# Abstract Class
class Payment(ABC):

    def __init__(self, amount):
        self.amount = amount

    @abstractmethod
    def process_payment(self):
        pass

# Child Class 1
class CreditCard(Payment):

    def __init__(self, amount):
        super().__init__(amount)   # Call parent constructor

    def process_payment(self):
        print(f"Credit Card Payment of processed : ₹{self.amount}")

    def display(self):
        print("Payment Method : Credit Card")
        print("Amount         :", self.amount)
        self.process_payment()

# Child Class 2
class Cash(Payment):

    def __init__(self, amount):
        super().__init__(amount)   # Call parent constructor

    def process_payment(self):
        print(f"Cash Payment of received : ₹{self.amount}")

    def display(self):
        print("Payment Method : Cash")
        print("Amount         :", self.amount)
        self.process_payment()
        
# Objects
credit = CreditCard(50000)
cash = Cash(1200000)

credit.display()
print()
cash.display()
```

```python
# 22. Class Name: BankAccount
# Problem
# Create an abstract class named BankAccount.
# Explanation
# Store the account balance.
# __init__(self, balance)
# Create abstract method:
# • calculate_interest()
# Create subclass:
# • SavingsAccount
# Use super() to initialize the balance.

from abc import ABC, abstractmethod

# Abstract Class
class BankAccount(ABC):

    def __init__(self, balance):
        self.balance = balance

    @abstractmethod
    def calculate_interest(self):
        pass

# Child Class
class SavingsAccount(BankAccount):

    def __init__(self, balance, interest_rate):
        super().__init__(balance)   # Call parent constructor
        self.interest_rate = interest_rate

    def calculate_interest(self):
        return (self.balance * self.interest_rate) / 100

    def display(self):
        print("Account Type  : Savings Account")
        print("Balance       :", self.balance)
        print("Interest Rate :", self.interest_rate, "%")
        print("Interest      :", self.calculate_interest())

# Object
account = SavingsAccount(500000, 8)

account.display()
```
