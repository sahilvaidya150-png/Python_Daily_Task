### Q19. Given data = {"a": 10, "b": 20, "c": 10}, use items() and print only keys where value is 10.

```python
data = {"a": 10, "b": 20, "c": 10}

for key, value in data.items():
    if value == 10:
       print(key)
```

```python
for key , value in data.items():
    if value ==10:
        print(key)
```

### Q21. Create a dictionary student with 4 key-value pairs. Then use update() to add another dictionary, remove the last item using popitem(), and finally print all items using items().

```python
student={
    "sahil":89,
    "piyush":88,
    "yash":87,
    "sanket":86
}
print(student)

# Then use update()

New_student=student.update({"sameer":85})
print("New update student:",student)

# remove the last item using popitem()

removed_item =student.popitem()
print("pop student:",removed_item)

# finally print all items using items().

for key , value in student.items():
    print(key ,":",value)
```


### Q22. Create a tuple with 8 numbers. Find the index of the number 25. If it exists, count how many times it appears. Then convert the tuple into a list, sort it in descending order, and print.

```python
tup_number= (10,20,25,40,50,60,25,25,80)

if 25 in tup_number:
    print("INdex number:",tup_number.index(25))
    print("Count number:",tup_number.count(25))

else:
    print("25 is not in present")

num_list=list(tup_number)
print("Tuple into a lsit:",num_list)

num_list.sort(reverse=True)
print("descending order:",num_list)
```

### Q23. Take 4 student names and their marks as input from the user. Store them in a dictionary (name as key, marks as value). Then find and print the student with the highest marks.

```python
student={}

for i in range(5):
    name=input("enter the name:")
    marks=int (input("enter the marks:"))
    student[name]=marks

print("Student Dictionary")
print(student)

highest_marks=max(student, key=student.get)

print("Student with highest marks:", highest_marks)
print("Highest marks:", student[highest_marks])
```

### Q24. Create a dictionary info = {"name": "Aman", "age": 21, "city":"Mumbai"}. Make a copy of it. Clear the original dictionary. Then add a new key "country": "India" in the copy and print both dictionaries.

```python
info = {"name": "Aman", "age": 21, "city":"Mumbai"}
print("original:",info)

original_copy=info.copy()
print("copy:",original_copy)

info.clear()
print("clear:",info)

new_key = original_copy.update({"country": "India"})
print("new key:",original_copy)
```

### Q25. Given data = (10, 20, 30, 20, 40, 20, 50), convert it into a dictionary where key is the number and value is its count of occurrences (using tuple methods + dictionary)

```python
data = (10, 20, 30, 20, 40, 20, 50)

number = {}

for num in data:
    number[num] = data.count(num)

print(number)
```
