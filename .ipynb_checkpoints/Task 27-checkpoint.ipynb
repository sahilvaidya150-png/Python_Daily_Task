#### 1. Take a list of numbers. Using loop and operators, separate even and odd numbers, then calculate sum of even and product of odd numbers without using built-in sum() or prod().

```python
numbers = [10, 15, 8, 7, 6, 3, 4]

even = []
odd = []

even_sum = 0
odd_product = 1

for num in numbers:
    if num % 2 == 0:
        even.append(num)
        even_sum += num
    else:
        odd.append(num)
        odd_product *= num

print("Even Numbers :", even)
print("Odd Numbers  :", odd)
print("Sum of Even Numbers :", even_sum)
print("Product of Odd Numbers :", odd_product)
```

#### 2. Create a program using nested loops and break/continue that prints all prime numbers between two user-given numbers, and also counts how many primes were found.

```python
start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

count = 0

print("Prime numbers are:")

for num in range(start, end + 1):
    if num < 2:
        continue   # Skip numbers less than 2

    for i in range(2, num):
        if num % i == 0:
            break   # Not a prime number
    else:
        print(num)
        count += 1

print("Total Prime Numbers:", count)
```

#### 3. Take a string from user. Using string methods and loop, count vowels, consonants, digits and special characters separately.

```python
text = input("Enter a string: ")

vowels = 0
consonants = 0
digits = 0
special = 0

for ch in text:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1

print("\nResults:")
print("Vowels            :", vowels)
print("Consonants        :", consonants)
print("Digits            :", digits)
print("Special Characters:", special)
```

#### 4. Create a class BankAccount with private __balance.
#### • Add deposit and withdraw with validation
#### • Maintain a class attribute total_accounts
#### • Add a class method to show total accounts
#### • Overload + operator to transfer money from one account to another

```python
class BankAccount:

    # Class attribute
    total_accounts = 0

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
        BankAccount.total_accounts += 1

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Invalid deposit amount.")

    # Withdraw method
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    # Show balance
    def show_balance(self):
        print(f"{self.name}'s Balance: ₹{self.__balance}")

    # Class method
    @classmethod
    def show_total_accounts(cls):
        print("Total Accounts:", cls.total_accounts)

    # Operator overloading (+) for money transfer
    def __add__(self, other):
        amount = int(input("Enter amount to transfer: "))

        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            other.__balance += amount
            print(f"₹{amount} transferred from {self.name} to {other.name}.")
        else:
            print("Transfer failed. Invalid amount or insufficient balance.")

        return other


# Create two accounts
acc1 = BankAccount("Sahil", 10000)
acc2 = BankAccount("piyush", 5000)

# Deposit
acc1.deposit(2000)

# Withdraw
acc1.withdraw(3000)

# Display balances
acc1.show_balance()
acc2.show_balance()

# Show total accounts
BankAccount.show_total_accounts()

# Transfer money using + operator
acc1 + acc2

# Updated balances
acc1.show_balance()
acc2.show_balance()
```

#### 9. Create a class ComplexNumber and overload +, -, *, and == operators completely.

```python
class ComplexNumber:

    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    # Overload + operator
    def __add__(self, other):
        return ComplexNumber(
            self.real + other.real,
            self.imag + other.imag
        )

    # Overload - operator
    def __sub__(self, other):
        return ComplexNumber(
            self.real - other.real,
            self.imag - other.imag
        )

    # Overload * operator
    def __mul__(self, other):
        real = (self.real * other.real) - (self.imag * other.imag)
        imag = (self.real * other.imag) + (self.imag * other.real)
        return ComplexNumber(real, imag)

    # Overload == operator
    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag

    # Display complex number
    def __str__(self):
        return f"{self.real} + {self.imag}i"


# Create objects
c1 = ComplexNumber(4, 3)
c2 = ComplexNumber(2, 5)

# Addition
print("Addition:", c1 + c2)

# Subtraction
print("Subtraction:", c1 - c2)

# Multiplication
print("Multiplication:", c1 * c2)

# Equality
print("Are Equal?", c1 == c2)

c3 = ComplexNumber(4, 3)
print("c1 == c3 ?", c1 == c3)
```
