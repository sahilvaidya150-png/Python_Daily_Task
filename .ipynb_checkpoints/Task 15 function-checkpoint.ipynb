### 16. Create a function calculate(a, b, operation) and show how keyword args improvereadability.

```python
def calculate(a, b, operation):
    if operation == "add":
        return a + b
    elif operation == "sub":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        if b != 0:
            return a / b
        else:
            return "Cannot divide by zero"
    else:
        return "Invalid operation"

calculate(a=12,b=23,operation="add")
```

### 17. Write a function order(item, quantity, price) and call it with mixed positional + keyword.

```python
def order(item, quantity, price):
    return f"item={item}, \nquantity={quantity}, \nprice={price}"

order("bottle",100,20)
```

### 18. What is the advantage of using keyword arguments over positional?

```python
# The main advantage of keyword arguments is that they make your code more readable, flexible, and less prone to mistakes.

# 1. Better Readability

# With keyword arguments, it's clear what each value represents.

# Positional Arguments:

# def calculate(a, b, operation):
#     return f"{operation}: {a}, {b}"

# calculate(10, 5, "add")

# It is not immediately obvious what 10, 5, and "add" mean.

# Keyword Arguments:

# calculate(a=10, b=5, operation="add")

# Now it is clear that:

# a = 10
# b = 5
# operation = "add"
# 2. Order Does Not Matter

# Positional Arguments:

# calculate(10, 5, "add")   # Correct

# Changing the order gives the wrong result.

# Keyword Arguments:

# calculate(operation="add", b=5, a=10)

# This works because Python matches arguments by name, not by position.

# 3. Fewer Mistakes

# Suppose a function has many parameters:

# def student(name, age, city):
#     print(name, age, city)

# Using positional arguments:

# student("Sahil", 22, "Nagpur")

# If you accidentally swap values:

# student("Sahil", "Nagpur", 22)

# The output will be incorrect.

# Using keyword arguments:

# student(name="Sahil", age=22, city="Nagpur")

# The values are assigned to the correct parameters.

# 4. Easier to Understand Long Function Calls
# send_email(
#     to="abc@example.com",
#     subject="Meeting",
#     message="See you tomorrow"
# )

# This is much easier to understand than:

# send_email("abc@example.com", "Meeting", "See you tomorrow")
# Summary
# Positional Arguments	Keyword Arguments
# Order matters	Order does not matter
# Less readable	More readable
# Easier to make mistakes	Less chance of mistakes
# Best for simple functions	Best for functions with many parameters

# In short: Keyword arguments make your code clearer, more flexible, and easier to maintain, especially when functions have multiple parameters.
```

### 19. Write a function add_all(*args) that returns sum of all numbers passed.

```python
def add_all(*num):
    c=sum(num)
    return c

add_all(1,4,5,7,8,9,5,5,8,5,58,8)
```

### 21. Write a function find_max(*args) that returns the maximum value.

### 22. Make a function concatenate(*args) that joins all string arguments with space.

```python
def concatenate(*args):
    return " ".join(args)


concatenate("sahil","piyush","sameer","yash","sanket")    
```

### 23. Write a function multiply(*args) that multiplies all numbers.

```python
def multiply(*args):
    mul=1
    for i in args:
        mul*=i
    return mul

multiply(2,5)
```

### 24. Create a function that takes student marks as *args and returns average.

```python
def student_marks(*args):
    return sum(args)/ len(args)

student_marks(98,87,76,65)
```

### 25. Write a function info(*args) that prints type and value of each argument.

```python
def info(*args):
    for arg in args:
        print(f"Type: {type(arg)}, Value: {arg}")

info("Sahil", 22, 95.5, True, [1, 2, 3])
```
