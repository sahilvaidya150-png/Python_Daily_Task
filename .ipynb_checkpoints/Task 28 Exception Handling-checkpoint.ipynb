#### Basic Exception Handling (1-10)
#### 1. Write a program that takes two numbers and performs division. Handle the ZeroDivisionError.

```python
try :
    a = 23
    b = 0
    c = a / b
    print(c)

except ZeroDivisionError:
    print("Zero Division Error.")
```

#### 2. Take a number from user and convert it to integer. Handle ValueError if user enters a string.

```python
try:
    num = int(input("Enter the number:"))
    print(num)
except ValueError:
    print("Value Error")
```

#### 3. Access an element from a list using index. Handle IndexError.

```python
try :
    a = [1,2,3,4,5,6,7,8,9]
    print(a)
    a[10]
    print(a)

except IndexError:
    print("Error : Index Error")
```

#### 4. Try to open a file that does not exist. Handle FileNotFoundError.

```python
try:
    file = open("data.txt", "r")
    content = file.read()
    print(content)
    file.close()

except FileNotFoundError:
    print("Error: The file does not exist.")
```

#### 5. Create a dictionary and access a key that does not exist. Handle KeyError.

```python
student = {
    "name": "Sahil",
    "age": 22,
    "course": "Python"
}
try:
    print(student["marks"])   # Key does not exist

except KeyError:
    print("Error: The key does not exist in the dictionary.")
```

#### 6. Write a program using try-except to handle multiple exceptions (ValueError and ZeroDivisionError).

```python
try :
    a = 20
    b = 20.2
    c = a + b
    print(c)

except (ValueError and ZeroDivisionError) as e:
    print(e)
    
```

#### 7. Demonstrate the use of else block with try-except.

```python
try :
    a = int(input())
    print(a)

except ValueError:
    print("Error : Value Error:")

else :
    print("Conversion successful")
```

#### 8. Demonstrate the use of finally block with an example.

```python
# try except else finally

try :
    a = int(input())
    b = int(input())

    c = a + b

except Exception as e :
    print(e)

else :
    print(c)

finally :
    print("All Process done")
```

#### Multiple Exceptions & Specific Handling (11-18)
#### 11. Handle ZeroDivisionError and ValueError in the same program using multiple except blocks.

```python
try :
    a = 23
    b = 0
    
    c = a / b
    print(c)


except ValueError as e:
    print(e)
    print("Value Error.")

except ZeroDivisionError as e :
    print(e)
    print("Zero Division Error.")

except TypeError :
    print("Type Error")
```

#### 13. Demonstrate nested try-except blocks.

```python
try:
    print("Outer try")

    try:
        num = int(input("Enter a number:"))
        print(num)

    except ValueError:
        print("Invalid number!")

except Exception:
    print("Some error occurred.")

finally:
    print("Program Ended.")
```

#### Raising Exceptions & Custom Exceptions (19-25)
#### 19. Write a function that takes age and raises ValueError if age is less than 18.

```python
def check_age(age):
    if age < 18 :
        raise ValueError("Age must be 18 or above.")
    print("You are eligible")

try:
    age = int(input("Enter the Age"))
    check_age(age)

except Exception as e:
    print(e)
    print("Value Error")
```

#### 20. Create a custom exception called NegativeNumberError and raise it when user enters a negative number.

```python
class NegativeNumberError(Exception):
    pass

def NegativeNumber(num):
    if num < 0:
        raise NegativeNumberError("Negative Number Error")

    else:
        print("Positive Number")

try :
    num = int(input("Enter the number:"))
    NegativeNumber(num)

except NegativeNumberError as e:
    print(f"Error : {e}")
```

#### 21. Create a custom exception InsufficientBalanceError for a banking program.

```python
class InsufficientBalanceError(Exception):
    pass

def InsufficientBalance(amount):
    if amount < 0:
        raise InsufficientBalanceError("")
```
