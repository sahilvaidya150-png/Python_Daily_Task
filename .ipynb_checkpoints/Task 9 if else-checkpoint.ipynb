### 3. Write a program to check whether a triangle is:
### o Equilateral
### o Isosceles
### o Scalene

```python
A=int(input("enter a side 1:"))
B=int(input("enter a side 2:"))
C=int(input("enter a side 3:"))


if (A + B >= C) and (A + C >= B) and (B + C >= A):
    print("valid triangle")
    
    if (A == B) and (B == C):
        print("Equilateral")
        
    elif (A == B) or (B == C) or (A == C):
        print("Isosceles")
        
    else:
        print("Scalene")

else :
    print("invalide triangle")
```

### 2. Write a program to calculate income tax:
### o Income below ₹250000 → No tax
### o ₹250000–₹500000 → 5%
### o ₹500000–₹1000000 → 20%
### o Above ₹1000000 → 30%

```python
income= float(input("enter a salary:"))

if income <= 25000:
    print("No Tax")

elif income >= 250000 and income <= 500000:
    tax = income * 5 / 100
    print("Paid Tax 5%:",tax)

elif income >= 500000 and income <= 1000000:
    tax = income * 20 / 100
    print("Paid Tax 20%:",tax)

else :
    tax = income * 30 / 100
    print("Paid Tax 30%:",tax)
```

### 3. Write a program to determine the type of roots in a quadratic equation using discriminant.

```python
a = float(input("Enter value of a: "))
b = float(input("Enter value of b: "))
c = float(input("Enter value of c: "))

d = b**2 - 4*a*c

if d > 0:
    print("Two distinct real roots")
elif d == 0:
    print("Two equal real roots")
else:
    print("Two complex roots")
```

### 7. Write a program for an online shopping discount system:

### o ₹2000–₹5000 → 20% discount

```python
A =int(input("Enter the price:"))

if A >= 5000:
    discount= A * 30 / 100
    print("30 % discount:", discount)

elif A >= 2000 and A <= 5000:
    discount= A * 20 / 100
    print("20 % discount:", discount)

else:
    discount= A * 10 / 100
    print("10 % discount:", discount)
```
