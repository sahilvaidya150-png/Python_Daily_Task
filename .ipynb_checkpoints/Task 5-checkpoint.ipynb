### Q17. Create a list. Insert "Python" at index 0, then remove the last element using pop(), and print the length.

### Q18. Take 5 city names as input from user and store them in a list. Then print them in reverse order.

```python
city=[]

city.append(input("Enter a city name 1:"))
city.append(input("Enter a city name 2:"))
city.append(input("Enter a city name 3:"))
city.append(input("Enter a city name 4:"))
city.append(input("Enter a city name 5:"))

city.reverse()
print(city)
```

### Q20. Create a list of student marks. Sort it, find the highest and lowest mark using list methods, and print them.

```python
marks=[75, 92, 68, 85, 99, 78]

marks.sort()
print(marks)

print("sorted marks:",marks)

print("Lowest Mark:", marks[0])
print("Highest Mark:", marks[-1])
```
