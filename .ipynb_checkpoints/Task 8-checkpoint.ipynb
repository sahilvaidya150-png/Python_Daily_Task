### Q15. Take three sides of a triangle. Check if they form a valid triangle. If yes, also print if it is Equilateral, Isosceles, or Scalene.

```python
A=int(input("enter a side 1:"))
B=int(input("enter a side 2:"))
C=int(input("enter a side 3:"))


if (A + B >= C) and (A + C >= B) and (B + C >= A):
    print("valid triangle")
    if (A == B) and (B == C):
        print("Equilateral")
    elif (A == B) or (B == C) or (A == C):
        print("Isosceles")
    else:
        print("Scalene")

else :
    print("invalide triangle")
```

### Q16. Take student marks in 5 subjects. Calculate percentage and give grade: 90+ → A+, 80-89 → A, 70-79 → B, 60-69 → C, below 60 → Fail.

```python
s1=int(input("enter a marks 1:"))
s2=int(input("enter a marks 2:"))
s3=int(input("enter a marks 3:"))
s4=int(input("enter a marks 4:"))
s5=int(input("enter a marks5:"))

total = s1+s2+s3+s4+s5
print("Total Marks:",total)
percentage = total /5
print("Total Percentage:",percentage)

if percentage >= 90:
    print("grade A+:",percentage)

elif percentage >= 80:
    print("grade A:",percentage)

elif percentage >= 70:
    print("grade B:",percentage)

elif percentage >= 60:
    print("grade C:",percentage)

else:
    print("fail")
```

### Q19. Take a character. Check if it is uppercase, lowercase, digit, or special character.

```python
char=input("Enter the character:")

if char.isupper():
    print(f"uppercase:{char}")

elif char.islower():
    print(f"Lowercase:{char}")

elif char.isdigit():
    print(f"Digit:{char}")

else:
    print(f"Special character:{char}")
```

### Q20. Take three numbers. Print them in ascending order using if-elif-else (without using sort()).

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a <= b and b <= c:
    print(a, b, c)
    
elif a <= c and c <= b:
    print(a, c, b)
    
elif b <= a and a <= c:
    print(b, a, c)
    
elif b <= c and c <= a:
    print(b, c, a)
    
elif c <= a and a <= b:
    print(c, a, b)
    
else:
    print(c, b, a)
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a <= b:
    if b <= c:
        print(a, b, c)
    else:
        if a <= c:
            print(a, c, b)
        else:
            print(c, a, b)
else:
    if a <= c:
        print(b, a, c)
    else:
        if b <= c:
            print(b, c, a)
        else:
            print(c, b, a)
```
