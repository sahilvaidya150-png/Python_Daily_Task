```python
import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5]

Y1 = [10, 20, 15, 25, 30]
Y2 = [5, 15, 20, 18, 35]

plt.plot(X, Y1, label="Line 1")
plt.plot(X, Y2, label="Line 2")

plt.title("Two Lines on the Same Graph")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.legend()

plt.show()
```

#### 11.Create a pie chart showing the percentage of customers from 4 different cities.

```python
import matplotlib.pyplot as plt

cities = ["Dubai", "Delhi", "Bangalore", "Hyderabad"]
customers = [30, 25, 20, 25]

plt.pie(
    customers,
    labels=cities,
    autopct="%1.1f%%"
)

plt.title("Percentage of Customers by City")

plt.show()
```

#### 12.Create a plot containing two subplots:
#### o First subplot → line plot
#### o Second subplot → scatter plot

```python
import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.figure(figsize=(10, 4))

# First subplot - Line Plot
plt.subplot(1, 2, 1)
plt.plot(X, Y)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")

# Second subplot - Scatter Plot
plt.subplot(1, 2, 2)
plt.scatter(X, Y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.tight_layout()
plt.show()
```

#### 13.Create a graph for the mathematical function:
#### for values of x from -10 to 10.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10, 11)
y = x ** 2

plt.plot(x, y)

plt.title("Graph of y = x²")
plt.xlabel("X")
plt.ylabel("Y")

plt.grid()
plt.show()
```

#### 14.Create a graph containing three mathematical functions:
#### Use different markers/line styles and add a legend.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-10, 11)

y1 = x ** 2
y2 = x
y3 = x ** 3

plt.plot(x, y1, marker="o", linestyle="-", label="y = x²")
plt.plot(x, y2, marker="s", linestyle="--", label="y = x")
plt.plot(x, y3, marker="^", linestyle=":", label="y = x³")

plt.title("Three Mathematical Functions")
plt.xlabel("X")
plt.ylabel("Y")

plt.legend()
plt.grid()

plt.show()
```

#### 15.Create a complete business-style visualization using Matplotlib:
#### o X-axis → Month
#### o Y-axis → Sales
#### o Plot monthly sales
#### o Add title
#### o Add X/Y labels
#### o Add markers
#### o Add legend
#### o Customize figure size

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [120, 150, 135, 180, 200, 220,
         210, 240, 230, 260, 280, 300]

plt.figure(figsize=(12, 6))

plt.plot(
    months,
    sales,
    marker="o",
    linestyle="-",
    label="Monthly Sales"
)

plt.title("Monthly Sales Performance")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()

plt.grid(True)
plt.tight_layout()

plt.show()
```

#### 2. Plot two lines representing Product A and Product B sales for 6 months. Adda legend and different line styles.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

product_a = [100, 120, 150, 140, 180, 200]
product_b = [80, 110, 130, 160, 170, 190]

plt.plot(months, product_a, linestyle="-", marker="o", label="Product A")
plt.plot(months, product_b, linestyle="--", marker="s", label="Product B")

plt.title("Product A and Product B Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)

plt.show()
```

#### 3. Create a bar chart showing the revenue of 5 different products. Display therevenue value on top of each bar.

```python
import matplotlib.pyplot as plt

products = ["Product A", "Product B", "Product C", "Product D", "Product E"]
revenue = [50000, 75000, 60000, 90000, 80000]

bars = plt.bar(products, revenue)

plt.title("Revenue of 5 Products")
plt.xlabel("Products")
plt.ylabel("Revenue")

# Display revenue value on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(height),
        ha="center",
        va="bottom"
    )

plt.show()
```
