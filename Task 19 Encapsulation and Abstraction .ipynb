```python
# 1. Class Name: BankAccount
#  Problem
#  Create a class named BankAccount.
#  Explanation
#  A bank account stores money. The balance should not be directly accessible from outside
#  the class, so create a private attribute named __balance.
#  Implement the following methods:
#  • deposit(amount)
#  o Deposit money into the account.
#  o Amount must be greater than 0.
#  • withdraw(amount)
#   o Withdraw money.
#  o Withdrawal should happen only if:
#  ▪ amount > 0
# ▪ amount <= available balance


class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private attribute

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Deposit amount ")

    # Withdraw method
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")

    # Display balance
    def display_balance(self):
        print(f"Available Balance: ₹{self.__balance}")


# Create object
account = BankAccount(5000)

# Initial balance
account.display_balance()

# Deposit money
account.deposit(2000)
account.display_balance()

# Withdraw money
account.withdraw(1000)
account.display_balance()

# Invalid withdrawal
account.withdraw(7000)

# Invalid deposit
account.deposit(-500)
```

```python
# 16. Class Name: Shape
# Problem
# Create an abstract class Shape.
# Explanation
# Declare abstract methods
# • calculate_area()
# • calculate_perimeter()
# Create subclasses like
# • Circle
# • Rectangle
# Implement both methods.

from abc import ABC, abstractmethod

# Abstract Class
class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass


# Child Class - Circle
class Circle(Shape):

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius * self.radius

    def calculate_perimeter(self):
        return 2 * 3.14 * self.radius


# Child Class - Rectangle
class Rectangle(Shape):

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)

# Create Objects
c1 = Circle(5)
r1 = Rectangle(15, 6)

# Circle Output
print("Circle")
print("Area      :", c1.calculate_area())
print("Perimeter :", c1.calculate_perimeter())
print()
# Rectangle Output
print("Rectangle")
print("Area      :", r1.calculate_area())
print("Perimeter :", r1.calculate_perimeter())
```

```python
# 21. Class Name: BankAccount
# Problem
# Create abstract class BankAccount.
# Explanation
# Abstract method
# calculate_interest()
# Implement in
# • SavingsAccount
# • CurrentAccount
# Each account type should use a different interest calculation.

from abc import ABC, abstractmethod

# Abstract Class
class BankAccount(ABC):

    def __init__(self, balance):
        self.balance = balance

    @abstractmethod
    def calculate_interest(self):
        pass


# Child Class - Savings Account
class SavingsAccount(BankAccount):

    def calculate_interest(self):
        interest = self.balance * 0.04   # 4% interest
        return interest


# Child Class - Current Account
class CurrentAccount(BankAccount):

    def calculate_interest(self):
        interest = self.balance * 0.01   # 1% interest
        return interest

# Create Objects
savings = SavingsAccount(50000)
current = CurrentAccount(50000)

# Display Interest
print("Savings Account")
print("Balance  :", savings.balance)
print("Interest :", savings.calculate_interest())

print("\nCurrent Account")
print("Balance  :", current.balance)
print("Interest :", current.calculate_interest())
```

```python
# 25. Class Name: Shape3D
# Problem
# Create abstract class Shape3D.
# Explanation
# Abstract methods
# • volume()
# • surface_area()
# Implement in
# • Cube
# • Sphere
# • Cylinder
# Each uses different formulas.

from abc import ABC, abstractmethod

# Abstract Class
class Shape3D(ABC):

    @abstractmethod
    def volume(self):
        pass

    @abstractmethod
    def surface_area(self):
        pass


# Child Class - Cube
class Cube(Shape3D):

    def __init__(self, side):
        self.side = side

    def volume(self):
        return self.side ** 3

    def surface_area(self):
        return 6 * (self.side ** 2)


# Child Class - Sphere
class Sphere(Shape3D):

    def __init__(self, radius):
        self.radius = radius

    def volume(self):
        return (4 / 3) * 3.14 * (self.radius ** 3)

    def surface_area(self):
        return 4 * 3.14 * (self.radius ** 2)


# Child Class - Cylinder
class Cylinder(Shape3D):

    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def volume(self):
        return 3.14 * (self.radius ** 2) * self.height

    def surface_area(self):
        return 2 * 3.14 * self.radius * (self.radius + self.height)

# Create Objects
cube = Cube(8)
sphere = Sphere(6)
cylinder = Cylinder(4, 5)

# Cube
print("Cube")
print("Volume       :", cube.volume())
print("Surface Area :", cube.surface_area())

# Sphere
print("\nSphere")
print("Volume       :", sphere.volume())
print("Surface Area :", sphere.surface_area())

# Cylinder
print("\nCylinder")
print("Volume       :", cylinder.volume())
print("Surface Area :", cylinder.surface_area())
```
