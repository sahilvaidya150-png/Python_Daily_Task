## String and String Slicing Questions

### Question 5:

```python
# Take your full name as input from user.
# Print the following:
# 1. Full name in uppercase
# 2. Full name in lowercase
# 3. Length of your name
# 4. First 3 characters
# 5. Last 3 characters

name=input("Full Name:")

print(name.upper())
print(name.lower())
print(len(name))
print(name[0:3])
print(name[-3:])
```

### Question 6:

```python
# Take any sentence from user.
# Print:
# - First 10 characters
# - Last 10 characters
# - Characters from index 5 to 15
# - Every 2nd character of the sentence

char=input("Enter the characters:")

print(char[0:11])
print(char[-10:])
print(char[5:16])
print(char[::2])
```

### Question 7:

```python
# Take a word from user.
# Check if the word is a palindrome or not.
# (Example: "madam", "radar")

word=input("Enter the word:")

if word==word [::-1]:
    print("palindrome")

else:
    print("not palindrome")
```

### Question 9:

```python
# Take a password from user.
# Check if password is strong:
# - Length >= 8
# - Contains '@' or '#'
# Print "Strong Password" or "Weak Password"

password=input("Enter the password :")

if len(password) >= 8 and ('@' in password or '#'in password):
    print("the strong password")

else:
    print("weak password")
```

### 2. Logical Operators + User Input

```python
# Take a number from user.
# Check whether the number is:
# Positive AND Even
# OR
# Negative OR Odd
# Print appropriate message


num=int(input("Enter the number:"))

if num > 0 and num %2==0:
    print("the number is positive and even")

elif num < 0 or num % 2 != 0:
    print("the number is negative or odd")

else:
    print("the number is zero")
```

### 3. String Operations

```python
# Take your city name as input.
# Print the following:
# 1. City name in reverse order
# 2. Number of vowels in the city name (a,e,i,o,u)
# 3. City name without first and last character

city=str(input("Enter the city name :"))

reverse_city="".join(reversed(city))  # city[::-1]
print("reverse:",reverse_city         #print("Reverse city name:",city[::-1])

if city.lower() in "aeiou":
    print("it is a vowel")
    
else:
    print("it is not vowel")

print(city[1:-1])
```

### 4. String Slicing

```python
# Take a mobile number as input (10 digits).
# Print:
# First 3 digits (network code)
# Last 4 digits
# Middle 3 digits

number=input("Enter a 10 digit mobile number:")

print("first 3 digits:",number[:3])
print("last 4 digits:",number[-4:])
print("middle 3 digits:",number[3:6])
```
