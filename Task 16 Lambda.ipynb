### 30. Interview Challenge
### Given:
### students = [
### {"name":"Ram","marks":[70,80,90]},
### {"name":"Amit","marks":[90,92,94]},
### {"name":"Neha","marks":[60,65,70]},
### {"name":"Raj","marks":[88,84,90]}
### ]
### Using only lambda with built-in functions (map, filter, sorted, max, min, reduce whereappropriate):
### 1. Calculate each student's average marks.
### 2. Sort students by average marks (highest first).
### 3. Find the topper.
### 4. Filter students whose average is greater than 80.
### 5. Print only the names of the filtered students.

```python
students = [
    {"name": "Ram", "marks": [70, 80, 90]},
    {"name": "Amit", "marks": [90, 92, 94]},
    {"name": "Neha", "marks": [60, 65, 70]},
    {"name": "Raj", "marks": [88, 84, 90]}
]

# 1. Average marks
average_marks = list(map(lambda x: {"name": x["name"], "average": sum(x["marks"]) / len(x["marks"])},students))
print("Average Marks:", average_marks)

# 2. Sort by average
sorted_students = sorted(students,key=lambda x: sum(x["marks"]) / len(x["marks"]),reverse=True)
print("Sorted:", sorted_students)

# 3. Topper
topper = max(students,key=lambda x: sum(x["marks"]) / len(x["marks"]))
print("Topper:", topper)

# 4. Filter average > 80
filtered_students = list(filter(lambda x: sum(x["marks"]) / len(x["marks"]) > 80,students))
print("Filtered:", filtered_students)

# 5. Names of filtered students
names = list(map(lambda x: x["name"], filtered_students))
print("Names:", names)
```

### 1. Student Ranking Challenge (★★★★★)
### students = [
### {"name": "Ram","marks": {"Math": 80, "Python": 95, "SQL": 88}},
### {"name": "Amit","marks": {"Math": 92, "Python": 90, "SQL": 91}},
### {"name": "Neha","marks": {"Math": 85, "Python": 99, "SQL": 94}},
### {"name": "Raj","marks": {"Math": 78, "Python": 70, "SQL": 80}}
### ]
### Using only
### • lambda
### • map
### • filter
### • sorted
### • reduce
### • max
### • min
### Perform the following:
### 1. Calculate every student's average.
### 2. Sort by average descending.
### 3. Find topper.
### 4. Find lowest scorer.
### 5. Print only students whose average >90.
### 6. Print only names.
### 7. Find total marks of all students.

```python
from functools import reduce

students = [
    {"name": "Ram","marks": {"Math": 80, "Python": 95, "SQL": 88}},
    {"name": "Amit","marks": {"Math": 92, "Python": 90, "SQL": 91}},
    {"name": "Neha","marks": {"Math": 85, "Python": 99, "SQL": 94}},
    {"name": "Raj","marks": {"Math": 78, "Python": 70, "SQL": 80}}
]

# 1. Average
average_students = list(map(lambda x: {"name": x["name"],"average": sum(x["marks"].values()) / len(x["marks"])},students))
print("Average:", average_students)

# 2. Sort
sorted_students = sorted(students,key=lambda x: sum(x["marks"].values()) / len(x["marks"]),reverse=True)
print("Sorted:", sorted_students)

# 3. Topper
topper = max(students,key=lambda x: sum(x["marks"].values()) / len(x["marks"]))
print("Topper:", topper)

# 4. Lowest
lowest = min(students,key=lambda x: sum(x["marks"].values()) / len(x["marks"]))
print("Lowest:", lowest)

# 5. Average > 90
high_students = list(filter(lambda x: sum(x["marks"].values()) / len(x["marks"]) > 90,students))
print("Average > 90:", high_students)

# 6. Names
names = list(map(lambda x: x["name"], students))
print("Names:", names)

# 7. Total Marks
total_marks = reduce(lambda total, student: total + sum(student["marks"].values()),students,0)
print("Total Marks:", total_marks)
```
