### Q16. Take a sentence. Split into words, remove all words that start with vowel (a,e,i,o,u), then join remaining words.

```python
sent=input("Enter a sentence:")

words=sent.split()

result=[]
for word in words:
    if word[0].lower() not in"aeiou":
        result.append(word)

final_sentence=" ".join(result)

print("original sentence:", sent)
print("after removing vowel words:",final_sentence)
```

### Q20. Take a string. Use replace(), strip(), split(), and join() together to clean and reformat it as a proper sentence.

```python
string="  Hello---world---from---python  "

text=(string.replace("---"," "))
striped=(text.strip())
splited=(striped.split())
print(" ".join(splited))
```
