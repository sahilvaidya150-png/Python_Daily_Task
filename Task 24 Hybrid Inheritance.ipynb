```python
# 1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
# from both B and C).
# Concept
# Hybrid Inheritance (Combination of Single + Multiple Inheritance)
# What to do
# • Create parent class A.
# • Create class B inheriting from A.
# • Create another class C.
# • Create class D inheriting from both B and C.
# How to solve
# 1. Create class A.
# 2. Add one method.
# 3. Create class B(A).
# 4. Create class C.
# 5. Create class D(B, C).
# 6. Create an object of D.
# 7. Call methods from all parent classes.

# Parent Class
class A:
    def method_a(self):
        print("Method from Class A")

# Single Inheritance
class B(A):
    def method_b(self):
        print("Method from Class B")

# Another Parent Class
class C:
    def method_c(self):
        print("Method from Class C")

# Hybrid Inheritance
class D(B, C):
    def method_d(self):
        print("Method from Class D")

# Object
obj = D()

obj.method_a()
obj.method_b()
obj.method_c()
obj.method_d()
```

```python
# 2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
# Concept
# Real-world Hybrid Inheritance
# What to do
# • Person is the base class.
# • Employee inherits Person.
# • Student inherits Person.
# • TeachingAssistant inherits Employee and Student.
# How to solve
# Hierarchy
# Person
# / \
# Employee Student
# \ /
# TeachingAssistant
# Steps
# 1. Store name in Person.
# 2. Employee stores salary.
# 3. Student stores marks.
# 4. TeachingAssistant stores department.
# 5. Display all information.

class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, salary):
        Person.__init__(self, name)
        self.salary = salary

class Student(Person):
    def __init__(self, name, marks):
        Person.__init__(self, name)
        self.marks = marks

class TeachingAssistant(Employee, Student):
    def __init__(self, name, salary, marks, department):
        Employee.__init__(self, name, salary)
        Student.__init__(self, name, marks)
        self.department = department

    def display(self):
        print("Name :", self.name)
        print("Salary :", self.salary)
        print("Marks :", self.marks)
        print("Department :", self.department)

ta = TeachingAssistant("Sahil", 500000, 92, "Computer Science")
ta.display()
```

```python
# 3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
# Concept
# Hybrid Inheritance
# What to do
# • Vehicle is parent.
# • Car inherits Vehicle.
# • Electric is another parent.
# • ElectricCar inherits Car and Electric.
# How to solve
# Hierarchy
# Vehicle
# |
# Car Electric
# \ /
# ElectricCar
# Steps
# 1. Vehicle has start().
# 2. Car has drive().
# 3. Electric has battery().
# 4. ElectricCar accesses all methods.

class Vehicle:
    def start(self):
        print("Vehicle Started")

class Car(Vehicle):
    def drive(self):
        print("Car is Driving")

class Electric:
    def battery(self):
        print("Battery is Fully Charged")

class ElectricCar(Car, Electric):
    def details(self):
        print("Electric Car Ready")

obj = ElectricCar()

obj.start()
obj.drive()
obj.battery()
obj.details()
```

```python
# 4. Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO).
# Concept
# Method Resolution Order (MRO)
# What to do
# Create
# A
# |
# B

# C

# D(B,C)
# Print
# D.mro()
# How to solve
# 1. Create four classes.
# 2. Give same method name in multiple classes.
# 3. Print ClassName.mro().
# 4. Observe Python's searching order.

class A:
    def show(self):
        print("Class A")

class B(A):
    def show(self):
        print("Class B")

class C:
    def show(self):
        print("Class C")

class D(B, C):
    pass

obj = D()

obj.show()

print(D.show())
```
