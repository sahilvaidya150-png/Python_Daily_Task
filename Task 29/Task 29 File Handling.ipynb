#### 4. Count the number of lines in a text file.

```python
file = open("detail.txt", "r")

lines = file.readlines() 
count = len(lines)

print("Number of lines =", count)

file.close()
```

#### 5. Count the number of words in a text file.

```python
file = open("detail.txt","r")

content = file.read()          # Read the entire file
words = content.split()        # Split text into words
count = len(words)             # Count the words
print("Number of words =", count)
file.close()
```

#### 6. Count the number of characters in a text file.

```python
file = open("detail.txt", "r")

content = file.read()          # Read the entire file
count = len(content)           # Count the characters
print("Number of characters =", count)

file.close()
```

#### 7. Read a file line by line using a loop.

```python
file = open("detail.txt", "r")

for line in file:
    print(line, end="")

file.close()
```

#### 8. Write a program to copy content from one file to another file.

```python
# Open the source file in read mode
source = open("detail.txt", "r")

# Read the content of the source file
content = source.read()

# Open the destination file in write mode
destination = open("copy.txt", "w")

# Write the content into the destination file
destination.write(content)

# Close both files
source.close()
destination.close()

print("File copied successfully!")
```

#### 9. Check if a file exists or not before reading it.

```python
import os

filename = "detail.txt"

if os.path.exists(filename):
    file = open(filename, "r")
    print(file.read())
    file.close()
else:
    print("File does not exist.")
```

#### 10. Write a program to delete a file (use os module).

```python
import os

filename = "detail.txt"

if os.path.exists(filename):
    os.remove(filename)
    print("File deleted successfully!")
else:
    print("File does not exist.")
```

#### 11. Take student name and marks from user and store them in a file.

#### 12. Read a file and print only the lines that contain the word “Python”.

```python
file = open("detail.txt", "r")

for line in file:
    if "Python" in line:
        print(line, end="")

file.close()
```

#### 13. Write a program to reverse the content of a file and save it in another file.

```python
file = open("detail.txt", "r")

content = file.read()

reverse_content = content[::-1]

new_file = open("reverse.txt", "w")
new_file.write(reverse_content)

file.close()
new_file.close()

print("Reversed content saved successfully!")
```

#### 14. Count how many times a particular word appears in a file.
