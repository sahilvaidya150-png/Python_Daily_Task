#### Basic Multiple Inheritance (1-10)
#### 1. Create two parent classes Father and Mother. Create a child class Child that inherits from both.

```python
# Parent class 1
class Father:
    def father_info(self):
        print("Father: Ramlal")
        print("Occupation: Engineer")

# Parent class 2
class Mother:
    def mother_info(self):
        print("Mother: Varsha")
        print("Occupation: Housewife")

# Child class inheriting from both Father and Mother
class Child(Father, Mother):
    def child_info(self):
        print("Child: Sahil")
        print("Class: Enginner")

# Create object
c = Child()

# Call methods from both parent classes and child class
c.father_info()
print()

c.mother_info()
print()

c.child_info()
```

#### 2. Define class A with method show() and class B with method display(). Create class C that inherits from both and calls both methods.

```python
# Parent class A
class A:
    def show(self):
        print("This is show() method from class A")

# Parent class B
class B:
    def display(self):
        print("This is display() method from class B")

# Child class C inheriting from A and B
class C(A, B):
    def call_methods(self):
        self.show()
        self.display()

# Create object of class C
obj = C()

# Call both parent class methods
obj.call_methods()
```

#### 3. Create classes Teacher and Author. Create class Person that inherits from both and has attributes from both.

```python
# Parent class 1
class Teacher:
    def __init__(self, subject):
        self.subject = subject

# Parent class 2
class Author:
    def __init__(self, book):
        self.book = book

# Child class inheriting from Teacher and Author
class Person(Teacher, Author):
    def __init__(self, name, subject, book):
        self.name = name
        Teacher.__init__(self, subject)
        Author.__init__(self, book)

    def display(self):
        print("Name   :", self.name)
        print("Subject:", self.subject)
        print("Book   :", self.book)

# Create object
p = Person("Sahil", "Python", "Learning Python")

# Display information
p.display()
```

#### 4. Define class Sports and class Academic. Create class Student that inherits from both.

```python
# Parent class 1
class Sports:
    def sports_info(self):
        print("Sport: Cricket")

# Parent class 2
class Academic:
    def academic_info(self):
        print("Subject: Data Science")

# Child class inheriting from Sports and Academic
class Student(Sports, Academic):
    def student_info(self):
        print("Name: Sahil")

# Create object
s = Student()

# Call methods
s.student_info()
s.sports_info()
s.academic_info()
```

#### 5. Create class Camera and class MusicPlayer. Create class Smartphone that inherits from both.

```python
# Parent class 1
class Camera:
    def take_photo(self):
        print("Camera: Taking a photo")

# Parent class 2
class MusicPlayer:
    def play_music(self):
        print("Music Player: Playing music")

# Child class inheriting from Camera and MusicPlayer
class Smartphone(Camera, MusicPlayer):
    def phone_info(self):
        print("Smartphone: Android Phone")

# Create object
phone = Smartphone()

# Call methods
phone.phone_info()
phone.take_photo()
phone.play_music()
```

#### 6. Define two classes with same method name. Create a child class that inherits from both and resolve the method conflict.

```python
# Parent class 1
class A:
    def show(self):
        print("show() method from class A")

# Parent class 2
class B:
    def show(self):
        print("show() method from class B")

# Child class inheriting from A and B
class C(A, B):
    def display(self):
        print("Calling show() method:")
        self.show()   # Calls A.show() because A is listed first

# Create object
obj = C()

# Call method
obj.display()
```

#### 7. Create class Engine and class Wheels. Create class Car that inherits from both.

```python
# Parent class 1
class Engine:
    def engine_info(self):
        print("Engine: 12000 CC Petrol Engine")

# Parent class 2
class Wheels:
    def wheels_info(self):
        print("Wheels: 4 Alloy Wheels")

# Child class inheriting from Engine and Wheels
class Car(Engine, Wheels):
    def car_info(self):
        print("Car: BMW M4")

# Create object
car = Car()

# Call methods
car.car_info()
car.engine_info()
car.wheels_info()
```

#### 8. Define class Writer and class Singer. Create class Artist that inherits from both.

```python
# Parent class 1
class Writer:
    def write(self):
        print("Writer: Writing a book")

# Parent class 2
class Singer:
    def sing(self):
        print("Singer: Singing a song")

# Child class inheriting from Writer and Singer
class Artist(Writer, Singer):
    def artist_info(self):
        print("Artist: Sahil")

# Create object
artist = Artist()

# Call methods
artist.artist_info()
artist.write()
artist.sing()
```

#### 9. Create class Bank and class Insurance. Create class Customer that inherits from both.

```python
# Parent class 1
class Bank:
    def bank_details(self):
        print("Bank: Bank Of India")
        print("Account Type: Savings")

# Parent class 2
class Insurance:
    def insurance_details(self):
        print("Insurance: Life Insurance")
        print("Policy No: INS12345")

# Child class inheriting from Bank and Insurance
class Customer(Bank, Insurance):
    def customer_info(self):
        print("Customer Name: Sahil")

# Create object
customer = Customer()

# Call methods
customer.customer_info()
customer.bank_details()
customer.insurance_details()
```

#### 10. Demonstrate Method Resolution Order (MRO) using multiple inheritance with three classes.

```python
# Parent class 1
class A:
    def show(self):
        print("show() method from class A")

# Parent class 2
class B(A):
    def show(self):
        print("show() method from class B")

# Parent class 3
class C(A):
    def show(self):
        print("show() method from class C")

# Child class
class D(B, C):
    pass

# Create object
obj = D()

# Call method
obj.show()

# Display Method Resolution Order
print(D.mro())
```
