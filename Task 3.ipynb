## String Slicing Questions

### Q1. Take a string from user and print the first 5 characters.

```python
name=input("Enter youyr name:")

print("5 characters:",name[0:5])
```

### Q2. Take a string and print the last 4 characters using negative slicing.


```python
name=input("Enter your name:")

print("4 characters:",name[-4:])
```

### Q3. Take a string and print characters from index 2 to 8 (excluding 8).

```python
text=input("Enter string:")

print("2 to 8 index:",text[2:8])
```

### Q4. Take a string and print every 2nd character of the string.

```python
text=input("Enter a string:")

print("Every 2nd char:",text[::2])
```

### Q5. Take a full name (e.g. "Sumit Kumar Sharma") and print only the first name and last name separately using slicing.

```python
name=input("Enter full name:")

print("first name:",name[0:6]) 
print("last name:",name[-6:])
```

## Case Conversion Questions

### Q6. Take a string from user and print it in:• Uppercase• Lowercase• Title Case• Capitalized (only first letter capital)

```python
text=input("Entetr a string:")

print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())
```

### Q7. Take a sentence and convert it to title case. Then print the original and converted version.

```python
text=input("Enter a sentence:")

title_case = text.title()
print("original:",text)
print("converted:", title_case)
```

### Q8. Take a mixed case string and check if lower() and upper() methods change the string permanently or not.

```python
text=input("Enter a mixed case string:")

print("orginal:",text)
print("lower case:",text.lower())
print("upper case",text.upper())

print("orginal string after method:",text)
```

### Q18. Take your full name.
### • Convert it to uppercase
### • Find the position of first space
### • Print the first name and last name separately.

```python
full_name="sahil ramlal vaidya"

print("uppercase:",full_name.upper())
print("first space:",full_name.find(" "))
print("first name:",full_name[:6]) 
print("last name:",full_name[-6:])
```

### Q19. Take a mobile number as string. Check if it is 10 digits and contains only numbers using isdigit() and len().

```python
number=input("Enter the number:")

if number.isdigit() and len(number)==10:
    print("its is a number")

else:
    print("its is a not number")
```
