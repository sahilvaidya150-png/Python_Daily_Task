```python
# Single-Level Inheritance (1–15)

# 1. Class Name: Person → Student
# Problem
# Create a parent class named Person and a child class named Student.
# Explanation
# The Person class should store:
# • name
# • age
# The Student class should inherit from Person and add:
# • marks
# Create a method in Student to display the student's name, age, and marks.


# Parent Class
class Person:

    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Child Class
class Student(Person):

    # Constructor
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    # Method to display details
    def display_info(self):
        print("----- Student Details -----")
        print(f"Name  : {self.name}")
        print(f"Age   : {self.age}")
        print(f"Marks : {self.marks}")

# Create Object
student1 = Student("Sahil", 23, 95)

# Call Method
student1.display_info()
```

```python
# 2. Class Name: Vehicle → Car
# Problem
# Create a parent class named Vehicle.
# Explanation
# The Vehicle class should contain a method:
# • start()
# Create a child class Car that:
# • Adds a brand attribute.
# • Overrides the start() method to display a car-specific starting message.
# Parent Class

class Vechicle:

    # Constructor
    def __init__(self, start):
        self.start = start
    
# Child Class
class Car(Vechicle):

    # Constructor
    def __init__(self, start, brand):
        super().__init__(start)
        self.brand = brand

    # Method to display details
    def display_info(self):
        print("----- Car Details -----")
        print(f"start  : {self.start}")
        print(f"Brand   : {self.brand}")

# Create Object
car = Car("start in key", "BMW")

# Call Method
car.display_info()  
```

```python
# Parent Class
class Vehicle:

    # Method
    def start(self):
        print("Vehicle is starting...")

# Child Class
class Car(Vehicle):

    # Constructor
    def __init__(self, brand):
        self.brand = brand

    # Overriding Parent Method
    def start(self):
        print(f"{self.brand} Car is starting with a push button.")
        
# Create Object
car1 = Car("BMW")

# Call Method
car1.start()
```

```python
# 4. Class Name: Employee → Manager
# Problem
# Create a parent class named Employee.
# Explanation
# The Employee class should store:
# • emp_id
# • salary
# Create child class Manager.
# Add:
# • department
# Display all employee information.

# Parent Class
class Employee:

    # Constructor
    def __init__(self, emp_id, salary):
        self.emp_id = emp_id
        self.salary = salary

# Child Class
class Manager(Employee):

    # Constructor
    def __init__(self, emp_id, salary, department):
        super().__init__(emp_id, salary)
        self.department = department

    # Method to display details
    def display_info(self):
        print("----- Employee -----")
        print(f"Emp ID     : {self.emp_id}")
        print(f"Salary     : {self.salary}")
        print(f"Department : {self.department}")

# Create Object
e1= Manager("Sahil123", 1000000, "Data Sci")

# Call Method
e1.display_info()    
```

```python
# 5. Class Name: Shape → Circle
# Problem
# Create a parent class named Shape.
# Explanation
# Create a method:
# • area()
# Create child class:
# • Circle
# Store:
# • radius
# Override area() to calculate the area of a circle.

class Shape:

    def area(self):
        print("area of shape")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        area = 3.14 * self.radius * self.radius
        print("Area of Circle:", area)

    def display_radius(self):
        print("Radius of Circle:", self.radius)

c = Circle(5)

c.display_radius()
c.area()
```

```python
# 6. Class Name: BankAccount → SavingsAccount
# Problem
# Create a parent class named BankAccount.
# Explanation
# Provide methods:
# • deposit()
# • withdraw()
# Create subclass:
# • SavingsAccount
# Add:
# • calculate_interest()

# Parent Class
class BankAccount:

    # Constructor
    def __init__(self, balance):
        self.balance = balance

    # Deposit Method
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited Amount : {amount}")

    # Withdraw Method
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdraw Amount : {amount}")
        else:
            print("Insufficient Balance")

# Child Class
class SavingsAccount(BankAccount):

    # Constructor
    def __init__(self, balance, interest_rate):
        super().__init__(balance)
        self.interest_rate = interest_rate

    # Calculate Interest
    def calculate_interest(self):
        interest = (self.balance * self.interest_rate) / 100
        print(f"Interest : {interest}")

# Create Object
account = SavingsAccount(1000000, 5)

# Deposit Money
account.deposit(2000)

# Withdraw Money
account.withdraw(3000)

# Calculate Interest
account.calculate_interest()

# Final Balance
print("Final Balance :", account.balance)
```

```python
# 7. Class Name: Product → Electronics
# Problem
# Create a parent class named Product.
# Explanation
# Store:
# • price
# Create method:
# • discount()
# Create child class:
# • Electronics
# Add:
# • warranty
# Display all information.

# Parent Class
class Product:

    # Constructor
    def __init__(self, price):
        self.price = price
        
   # discount method
    def discount(self, percent):
        discount_amount = (self.price * percent) / 100
        final_price = self.price - discount_amount

   # display method
        print(f"Original Price : ₹{self.price}")
        print(f"Discount ({percent}%) : ₹{discount_amount}")
        print(f"Final Price : ₹{final_price}")

# Child Class
class Electronics(Product):

    # Constructor
    def __init__(self, price, warranty):
        super().__init__(price)
        self.warranty = warranty

    # Display Method
    def display_info(self):
        print("----- Electronics Details -----")
        print(f"Price    : ₹{self.price}")
        print(f"Warranty : {self.warranty} Years")

# Create Object
mobile = Electronics(50000, 2)

# Display Details
mobile.display_info()

# Apply Discount
mobile.discount(10)
```

```python
# 8. Class Name: Book → TextBook
# Problem
# Create a parent class named Book.
# Explanation
# Store:
# • title
# • author
# Create subclass:
# • TextBook
# Add:
# • subject
# Display complete details.

# Parent Class
class Book:

    # Constructor
    def __init__(self, title, author):
        self.title = title
        self.author = author

# Child Class
class TextBook(Book):

    # Constructor
    def __init__(self, title, author, subject):
        super().__init__(title, author)
        self.subject = subject

    # Display Method
    def display_details(self):
        print("----- TextBook Details -----")
        print(f"Title   : {self.title}")
        print(f"Author  : {self.author}")
        print(f"Subject : {self.subject}")

# Create Object
book1 = TextBook("Python Programming", "Sahil Vaidya", "Data Science")

# Display Details
book1.display_details()
```

```python
# 9. Class Name: Device → Mobile
# Problem
# Create a parent class named Device.
# Explanation
# Create method:
# • turn_on()
# Create subclass:
# • Mobile
# Add method:
# • call(number)

# Parent Class
class Device:

    # Method
    def turn_on(self):
        print("Device is turned on.")

# Child Class
class Mobile(Device):

    # Method
    def call(self, number):
        print(f"Calling {number}...")
        
# Create Object
mobile = Mobile()

# Call Parent Class Method
mobile.turn_on()

# Call Child Class Method
mobile.call("1234567890")
```
