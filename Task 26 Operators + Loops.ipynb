#### 10. Create a dictionary of student marks. Use get(), keys(), values(), items(), update() and pop().

```python
student = {
    "Sahil": 95, 
    "Piyush": 94, 
    "Sameer": 93
}   # Create a dictionary of student marks

print("Original Dictionary:", student)

# get() - Get the value of a key
print("Marks of Sahil:", student.get("Sahil"))

# keys() - Display all keys
print("Keys:", student.keys())

# values() - Display all values
print("Values:", student.values())

# items() - Display all key-value pairs
print("Items:", student.items())

# update() - Add or update a key-value pair
student.update({"yash": 94})
print("After update():", student)

# pop() - Remove a key
student.pop("Sameer")
print("After pop():", student)
```

#### 11. Create two sets and perform union(), intersection(), difference(), add() and discard().

```python
sets1 = {1,2,3,4,5}
sets2 = {6,7,8,9,5}

print("sets 1:",sets1)
print("sets 2:",sets2)

# union()
print("Union:",sets1.union(sets2))

# intersection()
print("Intersection:",sets1.intersection(sets2))

# difference()
print("difference:",sets1.difference(sets2))

# add()
sets1.add(10)
print("Add:",sets1)

# discard()
sets1.discard(5)
print("Dicard:",sets1)
```

#### 14. Create class BankAccount with private attribute __balance. Add deposit and withdraw methods with validation.

```python
# Create BankAccount class
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # Private attribute

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount!")

    # Withdraw method
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount!")
        elif amount > self.__balance:
            print("Insufficient balance!")
        else:
            self.__balance -= amount
            print("Withdrawn:", amount)

    # Display balance
    def show_balance(self):
        print("Current Balance:", self.__balance)


# Create object
account = BankAccount(100000)

# Display initial balance
account.show_balance()

# Deposit money
account.deposit(500)
account.show_balance()

# Withdraw money
account.withdraw(300)
account.show_balance()

# Invalid withdrawal
account.withdraw(1500)     
```

#### 15. Create class Employee with private __salary. Add getter and setter. Setter should not allow salary below 5000.

```python
# Create Employee class
class Employee:
    def __init__(self, salary):
        self.__salary = salary   # Private attribute

    # Getter method
    def get_salary(self):
        return self.__salary

    # Setter method
    def set_salary(self, salary):
        if salary >= 5000:
            self.__salary = salary
            print("Salary updated successfully.")
        else:
            print("Salary cannot be less than 5000.")

# Create object
emp = Employee(100000)

# Get salary
print("Current Salary:", emp.get_salary())

# Update salary
emp.set_salary(15000)
print("Updated Salary:", emp.get_salary())

# Try to set salary below 5000
emp.set_salary(3000)
print("Final Salary:", emp.get_salary())
```
