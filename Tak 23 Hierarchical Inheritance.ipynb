```python
# 1. Person → Student, Teacher
# Question
# Create a parent class Person with attributes name and age. Create two child classes Student
# and Teacher that inherit from Person.
# Concept
# • Hierarchical Inheritance
# • Parent attributes inherited by multiple child classes.
# How to Solve
# 1. Create a parent class Person.
# 2. Add name and age using a constructor.
# 3. Create Student(Person).
# 4. Create Teacher(Person).
# 5. Create objects of both classes.
# 6. Show that both children can access name and age.

# Parent class
class Person:
    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Method to display details
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)

# Child class Student
class Student(Person):
    pass

# Child class Teacher
class Teacher(Person):
    pass

# Creating objects
student = Student("Rahul", 20)
teacher = Teacher("Anita", 35)

# Displaying details
print("Student Details:")
student.display()

print("\nTeacher Details:")
teacher.display()
```

```python
# 2. Animal → Dog, Cat
# Question
# Define class Animal with method eat(). Create subclasses Dog and Cat.
# Concept
# • Method inheritance.
# How to Solve
# 1. Create parent class Animal.
# 2. Write method eat().
# 3. Create child classes Dog and Cat.
# 4. Create objects.
# 5. Call eat() using both objects.

# Parent class
class Animal:
    def eat(self):
        print("Animal is eating")
        
# Child class Dog
class Dog(Animal):
    pass

# Child class Cat
class Cat(Animal):
    pass

# Creating objects
dog = Dog()
cat = Cat()

# Calling inherited method
dog.eat()
cat.eat()
```

```python
# 3. Vehicle → Car, Bike
# Question
# Create class Vehicle with method start(). Create subclasses Car and Bike.
# Concept
# • One parent method shared by multiple children.
# How to Solve
# 1. Create Vehicle.
# 2. Add start().
# 3. Inherit it in Car.
# 4. Inherit it in Bike.
# 5. Call start() from both child objects.

# Parent class
class Vehicle:
    # Method
    def start(self):
        print("Vehicle is starting")

# Child class Car
class Car(Vehicle):
    pass

# Child class Bike
class Bike(Vehicle):
    pass

# Creating objects
car = Car()
bike = Bike()

# Calling inherited method
print("Car:")
car.start()

print("\nBike:")
bike.start()
```

```python
# 4. Shape → Circle, Rectangle
# Question
# Define class Shape with method area(). Create subclasses Circle and Rectangle.
# Concept
# • Parent provides common method.
# • Children may later override it.
# How to Solve
# 1. Create Shape.
# 2. Add area().
# 3. Create Circle.
# 4. Create Rectangle.
# 5. Call area().

# Parent class
class Shape:
    # Method
    def area(self):
        print("Area of shape")

# Child class Circle
class Circle(Shape):
    pass

# Child class Rectangle
class Rectangle(Shape):
    pass

# Creating objects
circle = Circle()
rectangle = Rectangle()

# Calling inherited method
print("Circle:")
circle.area()

print("\nRectangle:")
rectangle.area()
```

```python
# 5. Employee → Manager, Developer
# Question
# Create class Employee with attributes emp_id and salary. Create subclasses Manager and
# Developer.
# Concept
# • Attribute inheritance.
# How to Solve
# 1. Create parent constructor.
# 2. Store emp_id and salary.
# 3. Create two child classes.
# 4. Show inherited attributes.

# Parent class
class Employee:
    # Constructor
    def __init__(self, emp_id, salary):
        self.emp_id = emp_id
        self.salary = salary

    # Method to display details
    def display(self):
        print("Employee ID:", self.emp_id)
        print("Salary:", self.salary)

# Child class Manager
class Manager(Employee):
    pass

# Child class Developer
class Developer(Employee):
    pass

# Creating objects
manager = Manager(101, 500000)
developer = Developer(102, 400000)

# Displaying inherited attributes
print("Manager Details:")
manager.display()

print("\nDeveloper Details:")
developer.display()
```
