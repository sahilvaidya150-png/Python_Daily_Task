### 4. Reverse a Number
### Question
### Write a program to reverse a number using a while loop.
### Explanation
### Example:
### 1234 → 4321
### Steps:
### • Take last digit using % 10
### • Add digit into reverse number
### • Remove last digit using // 10### 

```python
num = 1234
rev = 0 

while num > 0:
    digit = num  % 10
    rev =rev * 10 + digit
    num = num // 10

print(rev)
```

### 5. Count Digits in a Number
### Question
### Write a program to count total digits in a number.
### Explanation
### Example:
### 56789 → 5 digits
### Steps:
### • Use loop until number becomes 0
### • Remove one digit each time
### • Increase counter by 1

```python
# Input number from user
original_number = int(input("Enter a number: "))

# Handle the edge case where the number itself is 0
number = abs(original_number)
if number == 0:
    count = 1
else:
    count = 0

# 1. Use loop until number becomes 0
while number > 0:
    # 2. Remove one digit each time
    number = number // 10
    
    # 3. Increase counter by 1
    count += 1

print(f"Total digits in {original_number} is: {count}")
```

### 6. Palindrome Number
### Question
### Write a program to check whether a number is palindrome or not.
### Explanation
### Palindrome means same from both sides.
### Example:
### 121 → Palindrome
### 1331 → Palindrome
### Steps:
### • Reverse the number
### • Compare with original number

```python
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print(original, "is a Palindrome")
else:
    print(original, "is not a Palindrome")
```

### 7. Armstrong Number
### Question
### Write a program to check whether a number is Armstrong number.
### Explanation
### Example:
### 153
### Calculation:
### 13 + 53 + 33 = 153
### If sum equals original number → Armstrong number### 

```python
num = int(input("Enter a number: "))

original = num
total = 0

while num > 0:
    digit = num % 10
    total = total + (digit ** 3)
    num = num // 10

if total == original:
    print(original, "is an Armstrong number")
else:
    print(original, "is not an Armstrong number")
```

### 8. Password Checker
### Question
### Write a program to continuously ask password until correct password is
### entered.
### Explanation
### • Store one correct password
### • Use while loop
### • Keep asking user input
### • Stop when password matches

```python
# 1. Store one correct password
password = "sahil1234"

# Initialize the user's guess with an empty string
guess = ""

# 2. Use a while loop to keep asking while the guess is incorrect
while guess != password:
    # 3. Keep asking for user input
    guess = input("Enter the password: ")
    
    # Optional: Provide feedback if the password is wrong
    if guess != password:
        print("Incorrect password. Please try again.")

# 4. Stop when the password matches
print("Access granted! Correct password entered.")

```

### 9. Guessing Game
### Question
### Write a program for number guessing game.
### Explanation
### • Store one secret number
### • Ask user to guess
### • If wrong → ask again
### • If correct → print success message

```python
# 1. Store one secret number
secret_number = 7

# Initialize the user's guess variable
user_guess = None

# 3. If wrong -> ask again (loop runs until guess matches secret)
while user_guess != secret_number:
    # 2. Ask user to guess (convert input string to integer)
    user_guess = int(input("Guess the secret number (between 1 and 10): "))
    
    # Check if the guess is incorrect to give feedback
    if user_guess != secret_number:
        print("Wrong guess! Try again.")

# 4. If correct -> print success message
print("Congratulations! You guessed the correct number.")
```

### 11. Fibonacci Series
### Question
### Write a program to print Fibonacci series.
### Explanation
### Series:
### 0 1 1 2 3 5 8 13 ...
### Each next number is sum of previous two numbers.

```python
a = 0
b = 1
n = int(input("Enter the number:"))

count = 0
while count < n:
    print(a,end=" ")
    term = a + b
    a= b
    b=term
    count += 1
```
