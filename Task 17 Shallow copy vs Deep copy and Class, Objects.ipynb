### 1. Create a list original = [1, 2, [3, 4], 5]. Make a shallow copy named shallow and change shallow[2][0] = 99. Print both lists and explain the result.

```python
import copy
list_original = [1, 2, [3, 4], 5]
shallow=copy.copy(list_original)
shallow[2][0]=99        # shallow

print("list_original:",list_original)
print("shallow        :",shallow)

# A shallow copy creates a new outer list, but it does not create new copies of nested objects (like the inner list [3, 4]).
```

### 2. Do the same as above but use deepcopy instead of shallow copy. Show the difference.

```python
list_original = [1, 2, [3, 4], 5]
deep=copy.deepcopy(list_original)
deep[2][0]=99
print("list_original:",list_original)
print("deep copy    :",deep)
```

### 7. Write a function manual_deep_copy(lst) that manually creates a deep copy of a nested list without using copy.deepcopy().

```python
def manual_deep_copy(lst):
    new_list = []

    for item in lst:
        if isinstance(item, list):
            # Recursively copy nested lists
            new_list.append(manual_deep_copy(item))
        else:
            new_list.append(item)

    return new_list


# Original nested list
original = [1, 2, [3, 4, [5, 6]], 7]

# Create a manual deep copy
copied = manual_deep_copy(original)

# Modify the copied list
copied[2][2][0] = 99

# Print both lists
print("Original List:", original)
print("Copied List  :", copied)
```

### 9. (Hard) Create a function that detects whether a copy is shallow or deep by modifying a nested element and checking the original.

```python
import copy

def detect_copy(original, copied):
    # Save the original value
    old_value = original[1][0]

    copied[1][0] = 999              # Modify the nested element in the copied list

    if original[1][0] == 999:         # Check whether the original changed
        print("This is a Shallow Copy.")
    else:
        print("This is a Deep Copy.")
        
    original[1][0] = old_value      # Original list

data = [10, [20, 30], 40]          # Original list

shallow = copy.copy(data)          # Shallow copy

deep = copy.deepcopy(data)         # Deep copy

print("Testing Shallow Copy:")
detect_copy(data, shallow)

print("\nTesting Deep Copy:")
detect_copy(data, deep)
```

### 5. Bank Account
### Attributes
### • account_holder
### • balance
### Methods
### • deposit()
### • withdraw()
### Withdraw should happen only if balance is sufficient.
### Level 2 (Conditions)

```python
class Bank_account:
     def __init__(self, account_holder, balance):
         self.account_holder = account_holder
         self.balance = balance
         
         
     def deposit(self, amount):
             if amount > 0:
                 self.balance +=amount
                 print(f"Deposited ₹{amount}. Updated Balance: ₹{self.balance}")
             else:
                  print("Invalid deposit amount")

     def withdraw(self, amount):
            if amount <= 0:
                 print("Invalid withdrawal amount")
            elif amount <= self.balance:
                self.balance -= amount
                print(f"Withdraw ₹{amount}. Remaining Balance: ₹{self.balance}")
            else:
                print("Insufficient balance")


account= Bank_account("Sahil",100000000)

account.deposit(20000)
account.withdraw(5000)
account.withdraw(8000)
```

### Booking should happen only if seats are available.

```python
class Movie_ticket:
     def __init__(self, movie_name,seats ):
         self.movie_name = movie_name
         self.seats = seats
         
     def book_ticket(self,number):
         if number <= self.seats:
             self.seats -= number
             print("Ticket Booked")
             print("Available Seats:", self.seats)
         else:
             print("Seats Not Available")

movie = Movie_ticket("Sahil", 50)

movie.book_ticket(10)
```

```python
class Library:
     def __init__(self, book_name,available_copies ):
         self.book_name = book_name
         self.available_copies = available_copies

     def borrow_book(self):
         
         if self.available_copies > 0:
            self.available_copies -= 1
            print("Book Borrowed")
            print("Available Copies:", self.available_copies)
             
         else:
             print("Book Not Available")

book = Library("trading", 2)

book.borrow_book()
book.borrow_book()
book.borrow_book()
```

### 12. Employee Bonus
### If salary is
### greater than 50000
### increase by 20%
### otherwise
### increase by 10%.

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def bonus(self):
        if self.salary > 50000:
            self.salary = self.salary + (self.salary * 20 / 100)
        else:
            self.salary = self.salary + (self.salary * 10 / 100)

        print("Updated Salary:", self.salary)

emp1 = Employee("Sahil", 60000)
emp1.bonus()

emp2 = Employee("Piyush", 40000)
emp2.bonus()
```
