#### Q15
#### Suppose the Balance column has missing values.
#### Replace them using the mean of the Balance column.

```python
# Replace missing values in Balance with the mean
df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

# Display the updated Balance column
print(df["Balance"])
```

#### Q16
#### Suppose the Age column has missing values.
#### Replace them using the median age.

```python
# Replace missing values in Age with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Display the updated Age column
print(df["Age"])
```
