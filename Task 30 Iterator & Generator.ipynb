#### Level 2: Intermediate Iterator Problems
#### Problem 6
#### Create a custom iterator that prints numbers from 1 to 10.

```python
class Number_iterator:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value

        else :
            raise StopIteration

number = Number_iterator()

for i in number:
    print(i)
```

#### Problem 8
#### Create a custom iterator that prints numbers in reverse order from 10 to 1.

```python
class ReverseIterator:
    def __init__(self):
        self.num = 10

    def __iter__(self):
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value

        else :
            raise StopIteration

number = Number_iterator()

for i in number:
    print(i)
```

#### Problem 9
#### Create a custom iterator that prints the characters of the word "GENERATOR" one by one.

```python
class Custom_iterator:
    def __init__(self):
        self.word = "GENERATOR"
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch

        else :
            raise StopIteration

words = Custom_iterator()

for i in words:
    print(i)
```

#### Problem 10

#### Level 4: Generator with Lists
#### Problem 16
#### Given the following list:
#### numbers = [10, 15, 20, 25, 30, 35]
#### Create a generator that yields only the even numbers.

```python
numbers = [10, 15, 20, 25, 30, 35]

def even_generator():
    for num in numbers:
        if num % 2 == 0:
            yield num

# Create generator object
even = even_generator()

# Print even numbers
for i in even:
    print(i)
```

#### Problem 17

```python
numbers = [1, 4, 7, 10, 13, 16, 19]
def odd_generator():
    for num in numbers:
        if num % 2 != 0:
            yield num

# Create generator object
odd = odd_generator()

for i in odd:
    print(i)
```

#### Problem 18
#### Given a list of names, create a generator that yields one name at a time.
#### names = ["Alice", "Bob", "Charlie", "David", "Emma"]

```python
names = ["Alice", "Bob", "Charlie", "David", "Emma"]

def names_generator():
    for name in names:
            yield name

# Create generator object
name = names_generator()

for i in name:
    print(i)
```

#### Problem 19
#### Given a list of numbers, create a generator that yields only numbers greater than 50.
#### numbers = [20, 55, 40, 90, 60, 15, 75]

```python
numbers = [20, 55, 40, 90, 60, 15, 75]

def greater_generator():
    for num in numbers:
        if num > 50:
            yield num

# Create generator object
grt = greater_generator()

for i in grt:
    print(i)
```

#### Problem 20
#### Create a generator that yields the length of each word in the following list.
#### words = ["Python", "Java", "AI", "Generator", "Iterator"]

```python
words = ["Python", "Java", "AI", "Generator", "Iterator"]

def length_generator():
    for word in words:
        yield len(word)

# Create generator object
lengths = length_generator()

for i in lengths:
    print(i)
```

#### Problem 21
#### Create a custom iterator that prints the first 15 natural numbers.

```python
class NaturalNumbers:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 15:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration

numbers = NaturalNumbers()

for i in numbers:
    print(i)
```
