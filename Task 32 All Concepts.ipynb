#### Operators + Loops (1-5)
#### 1. Take two numbers from user. Perform all arithmetic operations and also check which number is greater using comparison operators.

```python
num1 = int(input("Enter the number 1:"))
num2 = int(input("Enter the number 2:"))

# arithmetic operations
print("Addition =", num1 + num2)
print("Subtraction =", num1 - num2)
print("Multiplication =", num1 * num2)
if num2 != 0:
    print("Division =", num1 / num2)
    print("Floor Division =", num1 // num2)
    print("Modulus =", num1 % num2)

else:
    print("Division, Floor Division, and Modulus are not possible")
print("Exponent =", num1 ** num2)

# comparison operators.
print()
print("\nComparison Operations")
print("num1 > num2 :", num1 > num2)
print("num1 < num2 :", num1 < num2)
print("num1 == num2 :", num1 == num2)
print("num1 != num2 :", num1 != num2)
print("num1 >= num2 :", num1 >= num2)
print("num1 <= num2 :", num1 <= num2)

# greater number
print()
if num1 > num2:
    print("First number is greater")
elif num2 > num1:
    print("Second number is greater")
else:
    print("Both numbers are equal")
```

#### 2. Print all numbers from 1 to 50. Skip multiples of 5 using continue and stop completely when number 40 is reached using break.

```python
for i in range(1,51):
    if i == 40:
        break
    if i % 5 == 0:
        continue
    print(i)
```

#### 3. Take numbers from user until they enter negative number. Calculate sum and average using while loop.

```python
sum = 0
count = 0

while True:
    num = int(input("Enter a number: "))

    if num < 0:
        break

    sum = sum + num
    count = count + 1

if count > 0:
    average = sum / count
    print("Sum =", sum)
    print("Average =", average)
else:
    print("No positive numbers entered.")
```

#### 5. Find the factorial of a number using for loop and also using while loop.

```python
# for loop
num = int(input("Enter the number:"))

fact = 1
for i in range(1,num+1):
    fact = fact * i

print("Factorail:",fact)
```

```python
# while loop

num = int(input("Enter the number:"))
fact = 1
i =1
while i <= num:
    fact = fact * i
    i = i + 1

print("Factorail:",fact)
```
