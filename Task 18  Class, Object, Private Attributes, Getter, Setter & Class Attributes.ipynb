### 13. In class Circle, add class attribute pi = 3.14. Use it in area calculation method.

```python
class Circle:
    # Class attribute
    pi = 3.14

    # Constructor
    def __init__(self, radius):
        self.radius = radius

    # Method to calculate area
    def area(self):
        area = Circle.pi * self.radius ** 2
        print(f"Radius : {self.radius}")
        print(f"Area   : {area}")


# Create objects
c1 = Circle(5)
c2 = Circle(7)

# Calculate area
c1.area()
print("-"*20)
c2.area()
```

### 15. Show difference between class attribute and instance attribute with example of class Player.

```python
class Player:
    # Class Attribute (shared by all objects)
    game = "Cricket"

    def __init__(self, name, score):
    
        self.name = name
        self.score = score

    def display(self):
        print(f"Player Name : {self.name}")
        print(f"Score       : {self.score}")
        print(f"Game        : {Player.game}")
        print("-" * 30)


# Create objects
player1 = Player("Virat", 95)
player2 = Player("Rohit", 80)

# Display details
player1.display()
player2.display()

# Change class attribute
Player.game = "Football"

print("After changing class attribute:\n")

player1.display()
player2.display()

# Change instance attribute of only player1
player1.score = 120

print("After changing instance attribute of player1:\n")

player1.display()
player2.display()
```
