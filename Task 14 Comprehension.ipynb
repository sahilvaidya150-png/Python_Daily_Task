### 7. What does round(3.678, 2) return? What about round(3.5) and round(2.5)?

```python
 ### The round() function is used to round a number to the specified number of decimal places.

print(round(3.678, 2))
# Round 3.678 to 2 decimal places.
# The third decimal digit is 8 (greater than 5), so the second decimal digit is increased.
# Result: 3.68

print(round(3.5))
# Python uses Banker's Rounding (Round Half to Even).
# 3.5 is exactly halfway between 3 and 4.
# 4 is the nearest even number.
# Result: 4

print(round(2.5))
# 2.5 is exactly halfway between 2 and 3.
# 2 is the nearest even number.
# Result: 2
```

### 8. Calculate pow(3, 4) and pow(2, 3, 5) (with modulo). What is the purpose of the third argument?

```python
# The pow() function is used to calculate the power (exponent) of a number. 
# It can also perform modular exponentiation when a third argument is provided.

print(pow(3,4))   #(3^4 = 3 \times 3 \times 3 \times 3 = 81)
print(pow(2,3,5))

# step 1: Calculate the power.
# 2³ = 8
# Step 2: Find the remainder when divided by 5.
# 8 % 5 = 3
# So,
# pow(2, 3, 5) = 3
```

### 20.Explain why set([1,1,2,2,3]) returns {1,2,3}.

```python
set([1,1,2,2,3]) 

# The list contains duplicate values:
# [1, 1, 2, 2, 3]
# When the list is converted into a set:
# set([1, 1, 2, 2, 3])
# Python keeps only one copy of each value.
# So,
# 1 appears twice → kept once.
# 2 appears twice → kept once.
# 3 appears once → kept once.
```

```python
# The reversed() function returns a reverse iterator that accesses the elements of a sequence in reverse order.
# It does not return a list directly.


print(reversed([1,2,3,4]))
# reversed() returns a reverse iterator.
# It does not display the reversed elements directly.
# The iterator can be used in a loop or converted into another data type.

print(list([1,2,3,4]))
# reversed(numbers) creates a reverse iterator.
# list() converts that iterator into a list.
# The original list remains unchanged.
```

### 29.Explain how zip([1,2,3], ['a','b','c']) works. What if lists have different lengths?

```python
# The zip() function combines elements from two or more iterables position by position. 
# It returns a zip object (iterator), which can be converted into a list using list().

numbers=[1,2,3]
letters=['a','b','c']

result = list(zip(numbers, letters))
print(result)
# zip() pairs the elements having the same index.

numbers=[1,2,3,4]
letters=['a','b',]
result = list(zip(numbers, letters)) 
print(result)
# numbers has 4 elements.
# letters has 2 elements.
# zip() creates pairs only until the shorter list ends.
# The remaining elements (3 and 4) are ignored.
```
