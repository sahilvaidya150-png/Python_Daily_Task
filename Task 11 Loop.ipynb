### 15. Take a positive number and keep dividing it by 2 using while loop until it becomes less than 1. Count the steps.

```python
num=int(input("Enter the number:"))
count=0

while num >= 1:
    num = num / 2
    count= count + 1
    print(num)

print("total:",count)
```

### 16. Use while loop to reverse a number (e.g., 1234 → 4321).

```python
num = 1234
rev = 0 

while num > 0:
    digit = num  % 10
    rev =rev * 10 + digit
    num = num // 10

print(rev)
```

### 17. Keep taking input from user until they enter "stop" (using while loop).

### 18. Print the Fibonacci series up to 10 terms using while loop.

```python
a = 0
b = 1
count = 0

while count < 10:
    print(a)
    c = a + b
    a = b
    b = c
    count = count + 1
```

### 20. Create a simple menu-driven program using while loop (options: 1. Add, 2. Subtract, 3. Exit).

```python
while True:
    print("\n1. Add")
    print("2. Subtract")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Addition =", a + b)

    elif choice == 2:
        a = int(input("Enter first number: "))
        b = int(input("Enter second number: "))
        print("Subtraction =", a - b)

    elif choice == 3:
        print("Program exited")
        break

    else:
        print("Invalid choice")
```
